Module 4: Quality Diagnostics and Group-Based Imputation

```python
#Missing Value Assessment

# Count missing values in every column
missing_values = df.isnull().sum()

# Display only columns that contain missing values
missing_values[missing_values > 0]
```

```python
#Missing Value Percentage

# Calculate the percentage of missing values in each column
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Display only columns with missing values
missing_percentage[missing_percentage > 0]
```

Group-Based Median Imputation

```python
# Define the columns that require imputation
imputation_columns = [
    "avg_sat_score",
    "avg_act_score",
    "ap_ib_participation_pct",
    "postsecondary_enrollment_pct"
]

# Create group-based median values
group_medians = df.groupby(
    ["nces_locale", "control"],
    observed=True
)[imputation_columns].transform("median")

# Fill missing values using the corresponding group median
df[imputation_columns] = df[imputation_columns].fillna(group_medians)

print("Group-based median imputation completed.")
```

Verify Missing Value Imputation

```python
# Check remaining missing values after group-based imputation
remaining_missing = df[imputation_columns].isnull().sum()

remaining_missing
```

Handling Groups Without a Valid Median

```python
# Check the remaining records with missing values
df[df[imputation_columns].isnull().any(axis=1)][
    ["institution_name", "nces_locale", "control"] + imputation_columns
]
```

Fallback Imputation Using Location Median

```python
# Use the nces_locale median as a fallback for groups
# where the nces_locale + control median is unavailable

location_medians = df.groupby(
    "nces_locale",
    observed=True
)[imputation_columns].transform("median")

df[imputation_columns] = df[imputation_columns].fillna(location_medians)

print("Fallback location-based imputation completed.")
```

Final Missing Value Verification

```python
# Verify that no missing values remain in the imputation columns
df[imputation_columns].isnull().sum()
```

SUMMARY

The dataset was assessed for missing values using Pandas. Missing values were identified in SAT scores, ACT scores, AP/IB participation, and postsecondary enrollment. Group-based median imputation was applied using nces_locale and control, followed by a location-level median fallback for groups where a specific group median was unavailable. The final verification confirmed that no missing values remained in the four affected indicators.

Module 5: Outlier Surgery, Detect, Erase, Heal

```python
# Calculate the first quartile and third quartile
Q1 = df["student_teacher_ratio"].quantile(0.25)
Q3 = df["student_teacher_ratio"].quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Calculate the lower and upper outlier fences
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Fence:", lower_fence)
print("Upper Fence:", upper_fence)
```

Identifying Outlier

```python
# Create a Boolean mask for Student-Teacher Ratio outliers
outlier_mask = (
    (df["student_teacher_ratio"] < lower_fence) |
    (df["student_teacher_ratio"] > upper_fence)
)

# Count the outliers
outlier_count = outlier_mask.sum()

print("Number of Student-Teacher Ratio outliers:", outlier_count)
```

Erase Outliers Using Boolean Indexing

```python
# Convert detected Student-Teacher Ratio outliers to NaN
df.loc[outlier_mask, "student_teacher_ratio"] = np.nan

# Confirm the number of missing values created
print(
    "Student-Teacher Ratio values converted to NaN:",
    df["student_teacher_ratio"].isna().sum()
)
```

Heal Outliers Using District Median

```python
# Calculate the median Student-Teacher Ratio for each county
county_median = df.groupby("county")["student_teacher_ratio"].transform("median")

# Identify the rows where Student-Teacher Ratio was converted to NaN
outlier_indices = df["student_teacher_ratio"].isna()

# Use Boolean indexing to identify the affected records
affected_rows = df.loc[
    outlier_indices,
    ["institution_name", "county", "student_teacher_ratio"]
]

affected_rows.head()
```

Fancy Indexing to Restore Outlier Values

```python
# Store the row positions of the outliers
outlier_positions = np.where(outlier_indices)[0]

# Use NumPy fancy indexing to obtain the corresponding county medians
replacement_values = county_median.iloc[outlier_positions].to_numpy()

# Replace the NaN outlier values in place
df.iloc[outlier_positions, df.columns.get_loc("student_teacher_ratio")] = replacement_values

print("Outlier values successfully healed:", len(outlier_positions))
```

Verify Outlier Healing

```python
# Confirm that Student-Teacher Ratio no longer contains the NaN values
print(
    "Remaining missing Student-Teacher Ratio values:",
    df["student_teacher_ratio"].isna().sum()
)
```

SUMMARY

The Student Teacher Ratio was assessed using the IQR method. The analysis identified 28 outliers outside the calculated IQR fences. These values were converted to NaN using Boolean indexing and subsequently replaced using the corresponding county median through NumPy fancy indexing. The final verification confirmed that all 28 outlier values were successfully healed and no missing Student Teacher Ratio values remained.

Module 6: Inferential Statistics and Hypothesis Testing

```python
# Import libraries required for statistical testing
import numpy as np
from scipy import stats

print("SciPy statistical tools loaded successfully.")
```

Problem Statement 1: Graduation Disparity

Null Hypothesis, H0: There is no significant difference in graduation rates between Public and Private institutions.

Alternative Hypothesis, H1: There is a significant difference in graduation rates between Public and Private institutions.

Prepare Public and Private Institution Groups

```python
# Define Public and Private institution groups

public_controls = [
    "Public",
    "Public Charter"
]

private_controls = [
    "Private For-Profit",
    "Private Non-Profit",
    "Private Religious"
]

public_group = df[
    df["control"].isin(public_controls)
]["graduation_rate_pct"].dropna()

private_group = df[
    df["control"].isin(private_controls)
]["graduation_rate_pct"].dropna()

print("Public institutions:", len(public_group))
print("Private institutions:", len(private_group))

print("Public mean graduation rate:", public_group.mean())
print("Private mean graduation rate:", private_group.mean())
```

Independent Samples T-Test

```python
# Perform Welch's independent samples t-test
t_statistic, p_value = stats.ttest_ind(
    public_group,
    private_group,
    equal_var=False
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

Independent T-Test Interpretation

The independent samples t-test produced a t-statistic of approximately -14.94 and a p-value of approximately 8.42 × 10⁻⁴⁷. Using a significance level of 0.05, the p-value is below the threshold. Therefore, the null hypothesis of no difference in graduation rates between the defined Public and Private groups is rejected. The analysis indicates a statistically significant difference in graduation rates between the two groups in this dataset.

Create Urban, Suburban, and Rural Groups

```python
# Create broad location groups from the NCES locale categories

urban_locales = [
    "City - Large",
    "City - Midsize"
]

suburban_locales = [
    "Suburban - Large",
    "Suburban - Midsize"
]

rural_locales = [
    "Town - Fringe",
    "Town - Distant",
    "Rural - Fringe",
    "Rural - Remote"
]

# Convert the categorical column to string before applying the conditions
locale_text = df["nces_locale"].astype(str)

df["broad_locale"] = "Other"

df.loc[locale_text.isin(urban_locales), "broad_locale"] = "Urban"
df.loc[locale_text.isin(suburban_locales), "broad_locale"] = "Suburban"
df.loc[locale_text.isin(rural_locales), "broad_locale"] = "Rural"

# Check the number of institutions in each group
df["broad_locale"].value_counts()
```

One-Way ANOVA: Student-Teacher Ratio by Location

```python
# Extract Student-Teacher Ratio for each broad location
urban_ratio = df.loc[
    df["broad_locale"] == "Urban",
    "student_teacher_ratio"
].dropna()

suburban_ratio = df.loc[
    df["broad_locale"] == "Suburban",
    "student_teacher_ratio"
].dropna()

rural_ratio = df.loc[
    df["broad_locale"] == "Rural",
    "student_teacher_ratio"
].dropna()

# Calculate the mean Student-Teacher Ratio for each group
print("Urban mean:", urban_ratio.mean())
print("Suburban mean:", suburban_ratio.mean())
print("Rural mean:", rural_ratio.mean())
```

ANOVA Statistical Test

```python
# Perform the one-way ANOVA
f_statistic, anova_p_value = stats.f_oneway(
    urban_ratio,
    suburban_ratio,
    rural_ratio
)

print("F-statistic:", f_statistic)
print("P-value:", anova_p_value)
```

ANOVA Interpretation

The one-way ANOVA produced an F-statistic of 0.2183 and a p-value of 0.8039. Since the p-value is greater than the 0.05 significance level, the null hypothesis is not rejected. The analysis does not provide evidence of a statistically significant difference in mean Student-Teacher Ratios among Urban, Suburban, and Rural institutions in this dataset.

Bootstrap Confidence Interval for Mean Math Proficiency

```python
# Extract valid mathematics proficiency scores
math_scores = df["math_proficiency_pct"].dropna().to_numpy()

# Set the number of bootstrap iterations
n_iterations = 10000

# Create an array to store bootstrap means
bootstrap_means = np.empty(n_iterations)

# Set a seed for reproducibility
np.random.seed(42)

# Generate 10,000 bootstrap samples and calculate their means
for i in range(n_iterations):
    sample = np.random.choice(
        math_scores,
        size=len(math_scores),
        replace=True
    )
    bootstrap_means[i] = np.mean(sample)

# Calculate the 95% confidence interval
lower_ci = np.percentile(bootstrap_means, 2.5)
upper_ci = np.percentile(bootstrap_means, 97.5)

print("Original mean Math Proficiency:", np.mean(math_scores))
print("Number of bootstrap iterations:", n_iterations)
print("95% Confidence Interval:")
print("Lower:", lower_ci)
print("Upper:", upper_ci)
```

Bootstrap Analysis Interpretation

The bootstrap analysis used 10,000 resampled iterations to estimate the confidence interval for mean mathematics proficiency. The observed mean mathematics proficiency was approximately 60.01%. The 95% bootstrap confidence interval ranges from approximately 59.55% to 60.46%, indicating the range of values supported by the bootstrap sampling distribution at the 95% confidence level.

SUMMARY

The inferential analysis used an independent samples t-test, one-way ANOVA, and bootstrap resampling to statistically evaluate the education data. The t-test identified a statistically significant difference in graduation rates between the defined Public and Private institution groups, while the ANOVA did not identify a statistically significant difference in Student-Teacher Ratios across Urban, Suburban, and Rural institutions. The bootstrap analysis estimated a 95% confidence interval of 59.55% to 60.46% for mean mathematics proficiency using 10,000 iterations.
